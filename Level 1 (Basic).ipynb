# Data Cleaning and Preprocessing

```python
import pandas as pd
```

```python
sentiment = pd.read_csv(r"C:\Users\LAWAL JNR\Desktop\codveda tech\Data Set For Task\3) Sentiment dataset.csv")
```

```python
#pd.set_option("display.max_rows", None)
pd.reset_option("display.max_rows")
```

```python
sentiment
```

```python
sentiment.info()
```

```python
sentiment = sentiment.drop(columns=["Unnamed: 0", "Unnamed: 0.1"])
```

```python
sentiment
```

```python
sentiment['Text'] = sentiment['Text'].str.strip()
sentiment['Sentiment'] = sentiment['Sentiment'].str.strip()
sentiment['User'] = sentiment['User'].str.strip()
sentiment['Platform'] = sentiment['Platform'].str.strip()
sentiment['Country'] = sentiment['Country'].str.strip().str.upper()
```

```python
sentiment['Timestamp'] = pd.to_datetime(sentiment['Timestamp'])
```

```python
print(sentiment['Timestamp'].dtype)
```

```python
sentiment
```

```python
sentiment["Date"] = sentiment["Timestamp"].dt.date
```

```python
sentiment
```

```python
sentiment["Time"] = sentiment["Timestamp"].dt.time
```

```python
sentiment[["Timestamp", "Date", "Time"]].head(10)
```

```python
sentiment = sentiment.drop(columns=["Timestamp"])
```

```python
sentiment
```

```python
sentiment.columns
```

```python
sentiment = sentiment[['Text', 'Sentiment', 'Date', 'Year', 'Month', 'Day', 'Time', 
                       'Hour', 'User', 'Platform', 'Hashtags','Retweets', 'Likes','Country']]
```

```python
sentiment
```

```python
sentiment.duplicated().sum()
```

```python
sentiment = sentiment.drop_duplicates()
```

```python
sentiment
```

# Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sentiment2 = pd.read_csv(r"C:\Users\LAWAL JNR\Desktop\codveda tech\Data Set For Task\task\level 1\cleaned_sentiment_dataset.csv")
```

```python
sentiment2[["Hour", "Retweets", "Likes"]].describe()
```

```python
sentiment2.loc[:, ["Retweets", "Likes"]] = sentiment2 [["Retweets", "Likes"]].apply (pd.to_numeric, errors = "coerce")

statistics = pd.DataFrame({
    "Mean" : sentiment2 [["Likes", "Retweets"]].mean(),
    "STD" : sentiment2 [["Likes", "Retweets"]].std(),
    "Median" : sentiment2 [["Likes", "Retweets"]].median(),
    "Mode" : sentiment2 [["Likes", "Retweets"]].mode().iloc[0]
}).T
```

```python
statistics
```

```python
sentiment2["Likes"].plot.hist(title = "Distribution of Likes", xlabel = "Likes", ylabel = "Counts")
plt.show()
```

```python
sentiment2.boxplot(column="Retweets", by = "Platform",  grid = False)
plt.xlabel("Platform")
plt.ylabel("Retweets")
plt.title("Retweets Distribution by Platform")
plt.suptitle("")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Likes', y='Retweets', hue='Platform', data= sentiment2)
plt.title('Relationship: Likes vs. Retweets')
plt.show()
```

```python
correlations = sentiment2[["Retweets", "Likes"]].corr()
correlations
```

```python
sns.heatmap(correlations, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
sentiment2["Country"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Countries")
plt.show()
```

```python
sentiment2["Month"].value_counts().sort_index().plot(kind="line")
plt.title("Posts Over Months")
plt.show()
```
